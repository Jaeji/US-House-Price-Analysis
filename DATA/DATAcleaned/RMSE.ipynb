```python
path = '/content/drive/MyDrive/1_2568_AM260_Grp05/Project/DATAcleaned/Housing(log).csv'
```

```python
# 1) Import library ที่จำเป็น
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import numpy as np

# 2) อ่าน Dataset (แก้เป็นไฟล์ของคุณ)
data = pd.read_csv(path)

# 3) เลือกตัวแปรอิสระ (X) และตัวแปรตาม (y)
X = data[['area', 'bathrooms', 'stories', 'mainroad', 'basement',
          'hotwaterheating', 'airconditioning', 'parking', 'prefarea',
          'unfurnished']]    # ← ตัวแปรจากโมเดล Gretl ของคุณ

y = data['price_log']        # ← ตัวแปรตาม

# 4) สร้างโมเดลและฟิตข้อมูล
model = LinearRegression()
model.fit(X, y)

# 5) ทำนายค่า (prediction)
y_pred = model.predict(X)

# 6) คำนวณค่า RMSE
rmse = np.sqrt(mean_squared_error(y, y_pred))

# 7) แสดงผลลัพธ์
print("RMSE =", rmse)
```

```python
# 1) Import libraries
import pandas as pd
from scipy import stats

# 2) โหลดข้อมูล (เปลี่ยนชื่อไฟล์ตามของคุณ)
data = pd.read_csv(path)

# 3) ค่าสถิติพื้นฐานทุกตัวแปร
summary = pd.DataFrame({
    'Mean': data.mean(),
    'Median': data.median(),
    'Mode': data.mode().iloc[0],   # mode อาจมีหลายค่า → เลือกค่าแรก
    'Min': data.min(),
    'Max': data.max(),
    'Std Dev': data.std(),
    'Variance': data.var(),
})

print(summary)
```
