```python
path = '/content/drive/MyDrive/Housing.csv'
```

```python
import pandas as pd
# อ่านไฟล์
df = pd.read_csv(path)
```

```python
# ขนาดของตารางข้อมูล
print(df.shape)
```

```python
# สถิติพื้นฐานของข้อมูลเชิงตัวเลข
print(df.describe())
```

**แปลงข้อมูล yes/no → Binary (0/1)**

```python
binary_cols = ['mainroad', 'guestroom', 'basement',
               'hotwaterheating', 'airconditioning', 'prefarea']

for col in binary_cols:
    df[col] = df[col].map({'yes': 1, 'no': 0})
```

```python
# ข้อมูล 5 แถวแรกของตาราง
print(df.head())
```

**แปลงข้อมูลหมวดหมู่หลายค่า → Dummy (One-hot encoding)**

```python
df = pd.get_dummies(df, columns=['furnishingstatus'], drop_first=True)
```

```python
# ข้อมูล 5 แถวแรกของตาราง
print(df.head())
```

```python
df = df.astype(int)
```

**ทำความสะอาดชื่อคอลัมน์**

```python
df.columns = [c.strip().replace(" ", "_").replace("-", "_") for c in df.columns]
```

**ตรวจสอบความสัมพันธ์เบื้องต้น (EDA)**

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Heatmap ของค่าสหสัมพันธ์
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation between variables")
plt.show()
```

**ทดลองรัน Regression เบื้องต้น**

```python
print(df.dtypes)
```

```python
df = df.astype(int)
```

```python
import statsmodels.api as sm

X = df.drop('price', axis=1)
y = df['price']
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()
print(model.summary())
```
